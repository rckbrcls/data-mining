```python
csv_path = "data/disque100-primeiro-semestre-2026.csv"
```

```python
from damicore import estimate, run

csv_options = {
    "split": "columns",
    "delimiter": ";",
    "encoding": "utf-8-sig",
}

preview = estimate(csv_path, **csv_options)
display(preview.model_dump())

result = run(csv_path, **csv_options, output_dir="output/damicore-run")
display(result.membership)
display(result.clusters)
print(result.tree_newick)
display(result.distance_matrix.head(10))
```

```python
result.save("output/result/damicore-result")
result.close()
```
